```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering
```

```python
# Load dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add species column
df['species'] = iris.target

# Display first five rows
print(df.head())

# Drop species column
X = df.drop('species', axis=1)

print("\nDataset used for clustering:")
print(X.head())
```

The Iris dataset contains 150 flower samples with four numerical features:

Sepal Length,
Sepal Width,
Petal Length,
Petal Width

It also contains a species (target) column. Since clustering is an unsupervised learning technique, the species labels are removed before training the models.

```python
# Apply KMeans
kmeans = KMeans(n_clusters=3, random_state=42)

clusters = kmeans.fit_predict(X)

# Add cluster labels
df['KMeans_Cluster'] = clusters

print(df.head())
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    X.iloc[:,0],
    X.iloc[:,2],
    c=clusters,
    cmap='viridis'
)

plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("KMeans Clustering")

plt.show()
```

KMeans is an unsupervised machine learning algorithm that groups similar data points into a specified number of clusters.

The algorithm works by:

Selecting K random centroids.
Assigning each point to the nearest centroid.
Updating the centroid by calculating the mean of each cluster.
Repeating the process until the centroids no longer change.
Why KMeans is Suitable for the Iris Dataset
The Iris dataset naturally contains three groups of flowers.
The numerical features make distance calculations effective.
KMeans performs well on small datasets with clearly separated clusters.

```python
hierarchical = AgglomerativeClustering(n_clusters=3)
h_clusters = hierarchical.fit_predict(X)
df['Hierarchical_Cluster'] = h_clusters
print(df.head())
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    X.iloc[:,0],
    X.iloc[:,2],
    c=h_clusters,
    cmap='rainbow'
)

plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Hierarchical Clustering")

plt.show()
```

Hierarchical Clustering builds clusters in a tree-like structure called a dendrogram.

There are two approaches:

Agglomerative (Bottom-Up)
Divisive (Top-Down)

In this assessment, Agglomerative Clustering is used.

It works by:

Treating each data point as an individual cluster.
Merging the two closest clusters.
Continuing until the desired number of clusters is obtained.
Why Hierarchical Clustering is Suitable for the Iris Dataset
No random centroid initialization is required.
It reveals the hierarchical relationship among observations.
It works well for small datasets like Iris.
