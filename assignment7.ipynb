to import libraries

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

to load dataset

```python
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target)
print(X.head())
print(y.head())
```

explore the dataset

```python
print(X.shape)
print(X.info())
print(X.describe())
```

checking missing values

```python
print(X.isnull().sum())
```

data preprocessing

feature scaling

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

split the dataset

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)
```

Preprocessing:
Missing Values

The Breast Cancer dataset contains no missing values, so no data cleaning or imputation was required.

Feature Scaling

The dataset contains numerical features with different ranges. Feature scaling was performed using StandardScaler, which standardizes the features by removing the mean and scaling to unit variance.

Feature scaling is necessary because:

SVM is distance-based.
K-NN uses Euclidean distance.
Logistic Regression performs better with standardized features.

Decision Tree and Random Forest do not require scaling, but scaling was applied uniformly to ensure fair model comparison.

logistic regrtession

```python
lr = LogisticRegression(max_iter=500)
lr.fit(X_train, y_train)
lr_pred = lr.predict(X_test)
```

Logistic Regression predicts probabilities using the logistic (sigmoid) function and classifies observations into two classes.

It is suitable because breast cancer classification is a binary classification problem.

decision tree

```python
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
dt_pred = dt.predict(X_test)
```

Decision Tree splits the dataset into smaller subsets using decision rules based on feature values.

It is suitable because it is easy to interpret and can capture nonlinear relationships.

random forest

```python
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)
```

Random Forest is an ensemble method that builds multiple decision trees and combines their predictions through majority voting.

It is suitable because it reduces overfitting and generally provides high accuracy.

support vector machine

```python
svm = SVC()
svm.fit(X_train, y_train)
svm_pred = svm.predict(X_test)
```

Support Vector Machine finds the optimal hyperplane that maximizes the separation between classes.

It is suitable because it performs well on high-dimensional datasets like the breast cancer dataset.

KNN

```python
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
knn_pred = knn.predict(X_test)
```

KNN classifies a sample based on the majority class among its nearest neighbors.

It is suitable because similar tumors are expected to have similar feature values.

Evaluate models

```python
models = {
    "Logistic Regression": lr_pred,
    "Decision Tree": dt_pred,
    "Random Forest": rf_pred,
    "SVM": svm_pred,
    "KNN": knn_pred
}
results = []
for name, pred in models.items():
    results.append([
        name,
        accuracy_score(y_test, pred),
        precision_score(y_test, pred),
        recall_score(y_test, pred),
        f1_score(y_test, pred)
    ])
results_df = pd.DataFrame(results,
                          columns=["Model","Accuracy","Precision","Recall","F1 Score"])
results_df
```

confusion matrix

```python
print(confusion_matrix(y_test, rf_pred))
```

classification report

```python
print(classification_report(y_test, rf_pred))
```

Model Comparison

The five classification algorithms were compared using Accuracy, Precision, Recall, and F1-score.

Logistic Regression achieved excellent performance due to the nearly linearly separable nature of the dataset.
Decision Tree performed well but was more prone to overfitting.
Random Forest improved upon Decision Tree by combining multiple trees, resulting in better generalization.
Support Vector Machine (SVM) achieved one of the highest accuracies because it effectively separates classes in high-dimensional feature space.
k-Nearest Neighbors (k-NN) also performed well after feature scaling but can be sensitive to the choice of k and the distribution of the data.
